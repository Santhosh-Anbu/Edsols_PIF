```python
import sounddevice as sd
import queue
import json
import time

from vosk import Model, KaldiRecognizer
import Arm_Lib
```

```python
Arm = Arm_Lib.Arm_Device()

HOME = [90,135,0,0,90,10]

Arm.Arm_serial_servo_write6_array(HOME,1000)
```

```python
def red():
    print("red")
    pose1=[54,68,15,37,87,126]
    pose2=[99,49,44,31,96,116]
    pose3=[90,90,90,90,90,90]

    Arm.Arm_serial_servo_write6_array(pose1,800)

    for i in range(1):

        Arm.Arm_serial_servo_write6_array(pose2,500)
        time.sleep(0.6)

        Arm.Arm_serial_servo_write6_array(pose3,500)
        time.sleep(0.6)
```

```python
model = Model("vosk-model-small-en-us-0.15")

q = queue.Queue()

def callback(indata, frames, time, status):
    q.put(bytes(indata))

mic_index = 3
device_info = sd.query_devices(mic_index, 'input')
samplerate = int(device_info['default_samplerate'])

rec = KaldiRecognizer(
    model,
    samplerate,
    '["red"]'
)
```

```python
running = True

last_command = ""

def execute_command(text):

    global running
    global last_command

    text = text.lower().strip()

    if text == last_command:
        return

    last_command = text

    print("Command :", text)

    if "red" in text:
        red()
    else: 
        print("Unknown command")
        
```

```python
with sd.RawInputStream(
    device=mic_index,
    samplerate=samplerate,
    blocksize=4096,
    dtype='int16',
    channels=1,
    callback=callback):
    print("----------------------------------")
    print("Voice Control Started")
    print("----------------------------------")
   
    print("red")
    print("----------------------------------")

    while running:
        data = q.get()

        if rec.AcceptWaveform(data):
            result = json.loads(rec.Result())
            text = result.get("text", "")

            if text:
                print("🧠 Heard:", text)
                execute_command(text)

print("Listener stopped")
home()

del Arm
```
